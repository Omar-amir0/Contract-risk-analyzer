```python
import pymupdf

pdf_path = "../knowledge_base/sample.pdf"

doc = pymupdf.open(pdf_path)

pages = []

for page_number, page in enumerate(doc):
    page_text = page.get_text()

    pages.append({
        "page": page_number + 1,
        "text": page_text
    })

print("Extracted pages:", len(pages))
```

```python
for page in pages:
    print(f"\n--- Page {page['page']} ---\n")
    print(page["text"][:1000])
```

```python
total_characters = sum(len(page["text"]) for page in pages)

print("Total characters extracted:", total_characters)

for page in pages:
    print(
        f"Page {page['page']}: {len(page['text'])} characters"
    )
```

# cleaning with regex

```python
import re
import html

def clean_text(text):
    # Decode HTML entities
    text = html.unescape(text)

    # Replace multiple spaces with a single space
    text = re.sub(r'[ \t]+', ' ', text)

    # Remove excessive blank lines
    text = re.sub(r'\n{3,}', '\n\n', text)

    return text.strip()
```

```python
cleaned_pages = []

for page in pages:
    cleaned_text = clean_text(page["text"])

    cleaned_pages.append({
        "page": page["page"],
        "text": cleaned_text
    })

print("Cleaning completed!")
```

```python
print("BEFORE CLEANING:")
print(pages[1]["text"][:500])

print("\nAFTER CLEANING:")
print(cleaned_pages[1]["text"][:500])
```

# chunking

```python
def split_into_chunks(text, chunk_size=500, overlap=75):

    words = text.split()

    chunks = []

    start = 0

    while start < len(words):

        end = start + chunk_size

        chunk = " ".join(words[start:end])

        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
document_chunks = []

for page in cleaned_pages:

    chunks = split_into_chunks(page["text"])

    for chunk in chunks:

        document_chunks.append({
            "source": "sample.pdf",
            "page": page["page"],
            "text": chunk
        })

print("Total chunks:", len(document_chunks))
```

```python
for i, chunk in enumerate(document_chunks[:3]):

    print(f"\n--- Chunk {i + 1} ---")
    print("Source:", chunk["source"])
    print("Page:", chunk["page"])
    print("Text:", chunk["text"][:500])
```

# EMBEDDING USING MINILM-L6-V2

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
sample_text = document_chunks[0]["text"]

embedding = model.encode(sample_text)

print("Embedding:", embedding)
print("Dimensions:", embedding.shape)
```

```python
from sentence_transformers import util

sentence_a = "The supplier can terminate the agreement."

sentence_b = "The vendor has the right to end the contract."

sentence_c = "The weather is sunny today."

embeddings = model.encode([
    sentence_a,
    sentence_b,
    sentence_c
])

similarity_ab = util.cos_sim(embeddings[0], embeddings[1])
similarity_ac = util.cos_sim(embeddings[0], embeddings[2])

print("A vs B:", similarity_ab.item())
print("A vs C:", similarity_ac.item())
```

```python
chunk_texts = [chunk["text"] for chunk in document_chunks]

embeddings = model.encode(
    chunk_texts,
    show_progress_bar=True
)

print("Number of embeddings:", len(embeddings))
print("Dimensions:", embeddings.shape)
```

# database config

```python
import chromadb

client = chromadb.PersistentClient(
    path="../database/chroma_db"
)
```

```python
collection = client.get_or_create_collection(
    name="legal_knowledge"
)
```

```python
ids = [f"legal_chunk_{i}" for i in range(len(document_chunks))]

texts = [chunk["text"] for chunk in document_chunks]

metadatas = [
    {
        "source": chunk["source"],
        "page": chunk["page"]
    }
    for chunk in document_chunks
]

collection.add(
    ids=ids,
    documents=texts,
    embeddings=embeddings.tolist(),
    metadatas=metadatas
)

print("Stored chunks:", collection.count())
```

# DATABASE TEST FINALLY

```python

query = "What were the security risks associated with PHP2Shell?"

query_embedding = model.encode(query).tolist()

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

for i in range(len(results["documents"][0])):

    print(f"\n--- Result {i + 1} ---")
    print("Source:", results["metadatas"][0][i]["source"])
    print("Page:", results["metadatas"][0][i]["page"])
    print("Text:", results["documents"][0][i][:500])
```
